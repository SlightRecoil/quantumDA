# Gates
## Hadamard Gate
The Hadamard gate is used to adapt a qubit into a super poisition. It does this by converting a computational bit into an equal superposition using the matrix:
$$
H = \frac{1}{\sqrt{2}}
\begin{pmatrix}
1 & 1 \\
1 & -1
\end{pmatrix}
$$
